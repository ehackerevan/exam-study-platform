```python
from pathlib import Path
import json
import runpy

root = Path.cwd()
while not (root / 'scripts' / 'analyze_exam_frequency.py').is_file() and root != root.parent:
    root = root.parent
if not (root / 'scripts' / 'analyze_exam_frequency.py').is_file():
    raise FileNotFoundError('找不到 Study Pack Compiler 專案根目錄')
runpy.run_path(str(root / 'scripts' / 'analyze_exam_frequency.py'), run_name='__main__')
result_path = root / 'tasks' / 'exam_frequency_20261020.json'
result = json.loads(result_path.read_text(encoding='utf-8'))
print(f'結果檔：{result_path}')
for course in result['courses'].values():
    print(course['label'], '紀錄', course['question_record_count'], '題；分類覆蓋', course['topic_classification_coverage_percent'], '%；未分類', course['unclassified_question_count'], '題')
```

```python
for course in result['courses'].values():
    print('\n' + course['label'])
    print('觀念排名：')
    for row in course['concept_ranking'][:10]:
        print(row['rank'], row['concept'], row['question_count'], '題 /', row['exam_session_count'], '考期')
    print('法規排名：')
    for row in course['law_ranking'][:10]:
        print(row['rank'], row['law_or_rule'], row['question_count'], '題 /', row['exam_session_count'], '考期')
    print('資料品質：ID 重複', course['duplicate_question_id_count'], '；答案索引錯誤', course['invalid_answer_index_count'], '；題次衝突組', course['same_exam_subject_number_collision_group_count'])
```
